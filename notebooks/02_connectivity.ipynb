# Connectivity Lists and Base Pairs

This notebook demonstrates connectivity list generation and base pair operations:
- Basic connectivity lists
- Extended bracket notation (pseudo-knots)
- Alternative pairing formats (letters, numbers)
- Base pair operations


```python
from rna_secstruct import SecStruct, get_connectivity_list, ConnectivityList, STANDARD_BRACKET_TYPES
from rna_secstruct.connectivity import connectivity_list, has_pseudoknot
```

## Basic Connectivity Lists


```python
# Simple connectivity list
structure = "(((...)))"
conn = connectivity_list(structure)
print(f"Structure: {structure}")
print(f"Connectivity: {conn}")
print(f"\nPosition 0 pairs with position {conn[0]}")
print(f"Position 1 pairs with position {conn[1]}")
print(f"Position 3 is unpaired (value: {conn[3]})")
```

## Using ConnectivityList Class


```python
# Create ConnectivityList with sequence and structure
cl = get_connectivity_list("GGGAAACCC", "(((...)))")

print(f"Sequence: {cl.sequence}")
print(f"Structure: {cl.structure}")
print(f"\nIs position 0 paired? {cl.is_nucleotide_paired(0)}")
print(f"Position 0 pairs with: {cl.get_paired_nucleotide(0)}")
print(f"Base pair at position 0: {cl.get_basepair(0)}")
print(f"Pair type at position 0: {cl.get_pair_type(0)}")
```

## Extended Bracket Notation (Pseudo-knots)


```python
# Pseudo-knot structure using multiple bracket types
pk_structure = "(([[))]]"
pk_sequence = "GGAAUUCC"

# Use standard bracket types for pseudo-knot support
cl_pk = get_connectivity_list(
    pk_sequence,
    pk_structure,
    bracket_types=STANDARD_BRACKET_TYPES
)

print(f"Pseudo-knot structure: {pk_structure}")
print(f"Connectivity: {cl_pk.connections}")
print(f"\nPosition 0 (() pairs with position 3 ())")
print(f"Position 1 ([) pairs with position 6 (])")
```

## Alternative Pairing Formats


```python
# Letter-based pairing notation
letter_structure = "a b c c b a"
letter_sequence = "GGGAAACCC"

cl_letter = get_connectivity_list(letter_sequence, letter_structure, format="letter")
print(f"Letter structure: {letter_structure}")
print(f"Connectivity: {cl_letter.connections}")
print(f"Pair types: {cl_letter.pair_types}")
print(f"\nPosition 0 (a) pairs with position 5 (a)")
print(f"Pair type at 0: {cl_letter.get_pair_type(0)}")
```

```python
# Number-based pairing notation
number_structure = "1 1 2 2 3 3"
number_sequence = "GGAAUU"

cl_number = get_connectivity_list(number_sequence, number_structure, format="number")
print(f"Number structure: {number_structure}")
print(f"Connectivity: {cl_number.connections}")
print(f"Pair types: {cl_number.pair_types}")
```

## Auto-detection of Format


```python
# Auto-detect format
bracket_struct = "(((...)))"
letter_struct = "a b c c b a"
number_struct = "1 1 2 2"

cl1 = get_connectivity_list("GGGAAACCC", bracket_struct)
cl2 = get_connectivity_list("GGGAAACCC", letter_struct)
cl3 = get_connectivity_list("GGAA", number_struct)

print(f"Bracket format detected: {cl1.get_pair_type(0) == '('}")
print(f"Letter format detected: {cl2.get_pair_type(0) == 'a'}")
print(f"Number format detected: {cl3.get_pair_type(0) == '1'}")
```

## Base Pair Operations with SecStruct


```python
# Use connectivity methods on SecStruct
struct = SecStruct("GGGAAACCC", "(((...)))")

print(f"Connectivity list: {struct.connectivity}")
print(f"\nIs position 0 paired? {struct.is_paired(0)}")
print(f"Base pair at position 0: {struct.get_basepair(0)}")
print(f"\nNumber of base pairs: {struct.get_num_basepairs()}")
print(f"Number of unpaired nucleotides: {struct.get_num_unpaired()}")
```
