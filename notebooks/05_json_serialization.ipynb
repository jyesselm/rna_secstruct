# JSON Serialization

This notebook demonstrates JSON serialization capabilities:
- Converting to/from dictionaries
- JSON string serialization
- File I/O operations
- Custom JSON encoder


```python
from rna_secstruct import SecStruct
from rna_secstruct.json_encoder import SecStructJSONEncoder, dumps, loads
import json
```

## Dictionary Conversion


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Convert to dictionary
d = struct.to_dict()
print(f"Dictionary keys: {list(d.keys())}")
print(f"Sequence: {d['sequence']}")
print(f"Structure: {d['structure']}")

# Convert back from dictionary
struct2 = SecStruct.from_dict(d)
print(f"\nRound-trip successful: {struct.sequence == struct2.sequence}")
```

## JSON String Serialization


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Serialize to JSON string
json_str = struct.to_json()
print(f"JSON (compact): {json_str}")

# Serialize with indentation
json_str_pretty = struct.to_json(indent=2)
print(f"\nJSON (pretty):\n{json_str_pretty}")

# Deserialize from JSON string
struct2 = SecStruct.from_json(json_str)
print(f"\nDeserialized successfully: {struct.sequence == struct2.sequence}")
```

## JSON with Motifs


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Trigger parsing to include motifs
_ = struct.motifs

# Convert to dictionary (includes motifs)
d = struct.to_dict()
print(f"Dictionary includes motifs: {'motifs' in d}")
if 'motifs' in d:
    print(f"Number of motifs: {len(d['motifs'])}")
    print(f"First motif keys: {list(d['motifs'][0].keys())}")

# Serialize to JSON
json_str = struct.to_json(indent=2)
print(f"\nJSON includes motifs: {'motifs' in json.loads(json_str)}")
```

## Custom JSON Encoder


```python
# Use custom encoder directly
struct = SecStruct("GGGAAACCC", "(((...)))")

# Encode using custom encoder
json_str = dumps(struct, indent=2)
print(f"Encoded with custom encoder:\n{json_str[:100]}...")

# Can also use with standard json.dumps
json_str2 = json.dumps(struct, cls=SecStructJSONEncoder, indent=2)
print(f"\nUsing json.dumps with encoder: {len(json_str2)} chars")
```

## File I/O


```python
import tempfile
import os

struct = SecStruct("GGGAAACCC", "(((...)))")

# Save to file
with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix='.json') as f:
    temp_file = f.name

try:
    struct.to_json_file(temp_file, indent=2)
    print(f"Saved to file: {temp_file}")
    print(f"File exists: {os.path.exists(temp_file)}")
    
    # Load from file
    struct2 = SecStruct.from_json_file(temp_file)
    print(f"\nLoaded from file successfully")
    print(f"Sequence matches: {struct.sequence == struct2.sequence}")
    print(f"Structure matches: {struct.structure == struct2.structure}")
finally:
    if os.path.exists(temp_file):
        os.remove(temp_file)
        print(f"\nCleaned up temporary file")
```

## API Integration Example


```python
# Example: Preparing data for API
struct = SecStruct("GGGAAACCC", "(((...)))")

# Convert to dictionary for API
api_data = struct.to_dict()
print(f"API-ready data: {list(api_data.keys())}")

# Or use JSON string
api_json = struct.to_json()
print(f"\nJSON string length: {len(api_json)} chars")

# Simulate receiving data from API
received_data = json.loads(api_json)
struct_received = SecStruct.from_dict(received_data)
print(f"\nReceived and reconstructed: {struct.sequence == struct_received.sequence}")
```
