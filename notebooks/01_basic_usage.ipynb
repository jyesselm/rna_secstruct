# RNA Secondary Structure - Basic Usage

This notebook demonstrates the basic functionality of the `rna_secstruct` library, including:
- Creating SecStruct objects
- Lazy loading behavior
- Basic properties and methods


```python
from rna_secstruct import SecStruct
import time
```

## Creating Structures


```python
# Create a simple hairpin structure
struct = SecStruct("GGGAAACCC", "(((...)))")
print(f"Sequence: {struct.sequence}")
print(f"Structure: {struct.structure}")
print(f"Length: {len(struct)}")
```

## Lazy Loading

Parsing is deferred until motifs are actually accessed, improving performance for large structures.


```python
# Create structure - parsing hasn't happened yet
struct = SecStruct("GGGAAACCC", "(((...)))")
print("Structure created (not parsed yet)")

# Accessing sequence/structure doesn't trigger parsing
_ = struct.sequence
_ = struct.structure
print("Accessed sequence/structure (still not parsed)")

# Accessing motifs triggers parsing
start = time.time()
motifs = struct.motifs
parse_time = time.time() - start
print(f"Accessed motifs (parsing triggered, took {parse_time:.4f}s)")
print(f"Number of motifs: {len(motifs)}")
```

## Basic Operations


```python
# String representation
print(f"String representation: {struct}")
print(f"Repr: {repr(struct)}")

# Length
print(f"Length: {len(struct)}")

# Slicing (returns new SecStruct)
sliced = struct[3:6]
print(f"Sliced [3:6]: {sliced.sequence}, {sliced.structure}")

# Original unchanged (immutable)
print(f"Original unchanged: {struct.sequence}")
```

## Combining Structures


```python
# Add two structures together
struct1 = SecStruct("GGG", "(((")
struct2 = SecStruct("AAA", "...")
combined = struct1 + struct2
print(f"Combined: {combined.sequence}, {combined.structure}")

# Join with strand separator
joined = struct1.join(struct2)
print(f"Joined: {joined.sequence}, {joined.structure}")
```

## Accessing Motifs


```python
# Get motifs (triggers parsing if not already done)
struct = SecStruct("GGGAAACCC", "(((...)))")
motifs = struct.motifs

print(f"Number of motifs: {len(motifs)}")
print(f"Motif IDs: {list(motifs.keys())}")

# Access specific motif
motif_0 = struct[0]  # Access by ID
print(f"\nMotif 0:")
print(f"  Type: {motif_0.m_type}")
print(f"  Sequence: {motif_0.sequence}")
print(f"  Structure: {motif_0.structure}")

# Iterate over motifs
print("\nAll motifs:")
for m_id, motif in struct.itermotifs():
    print(f"  {m_id}: {motif.m_type} - {motif.sequence}")
```

## Getting Motif Types


```python
# More complex structure
complex_struct = SecStruct("GGAAACGAAACGAAACC", "((...)(...)(...))")

print(f"Hairpins: {len(complex_struct.get_hairpins())}")
print(f"Helices: {len(complex_struct.get_helices())}")
print(f"Junctions: {len(complex_struct.get_junctions())}")
print(f"Single strands: {len(complex_struct.get_single_strands())}")
print(f"Total motifs: {complex_struct.get_num_motifs()}")
```
