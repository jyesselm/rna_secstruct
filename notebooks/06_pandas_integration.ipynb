# Pandas Integration

This notebook demonstrates pandas integration with RNA secondary structures:
- DataFrame and Series accessors
- Creating SecStruct columns from sequences/structures
- Vectorized operations
- JSON serialization with pandas


```python
import pandas as pd

```

## Creating DataFrames with RNA Structures


```python
# Create DataFrame with sequences and structures
df = pd.DataFrame({
    'name': ['hairpin1', 'hairpin2', 'junction1'],
    'sequence': ['GGGAAACCC', 'AAAGGGCCC', 'GGAAACGAAACGAAACC'],
    'structure': ['(((...)))', '(((...)))', '((...)(...)(...))']
})

print("Original DataFrame:")
print(df)

# Create SecStruct column using accessor
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')
print("\nDataFrame with SecStruct column:")
print(df[['name', 'secstruct']])
print(f"\nType of secstruct column: {type(df['secstruct'].iloc[0])}")
```

## Using the DataFrame Accessor


```python
# Add SecStruct column directly
df = pd.DataFrame({
    'sequence': ['GGGAAACCC', 'AAAGGGCCC'],
    'structure': ['(((...)))', '(((...)))']
})

df = df.rna.add_secstruct('sequence', 'structure', column='secstruct')
print("DataFrame with added SecStruct column:")
print(df.columns.tolist())
```

## Adding Statistics


```python
df = pd.DataFrame({
    'sequence': ['GGGAAACCC', 'AAAGGGCCC', 'CCCGAAAGGG'],
    'structure': ['(((...)))', '(((...)))', '(((...)))']
})
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')

# Add multiple statistics at once
df = df.rna.add_statistics('secstruct')

print("DataFrame with statistics:")
print(df[['sequence', 'secstruct_num_bp', 'secstruct_num_unpaired',
          'secstruct_gc_content', 'secstruct_length']])
```

## Series Accessor Methods


```python
df = pd.DataFrame({
    'sequence': ['GGGAAACCC', 'AAAGGGCCC'],
    'structure': ['(((...)))', '(((...)))']
})
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')

# Use Series accessor methods
num_bp = df['secstruct'].rna.num_basepairs()
num_motifs = df['secstruct'].rna.num_motifs()
gc_content = df['secstruct'].rna.gc_content()
helix_lengths = df['secstruct'].rna.helix_lengths()

print(f"Number of base pairs: {num_bp.tolist()}")
print(f"Number of motifs: {num_motifs.tolist()}")
print(f"GC content: {gc_content.tolist()}")
print(f"Helix lengths: {helix_lengths.tolist()}")
```

## JSON Serialization with Pandas


```python
df = pd.DataFrame({
    'sequence': ['GGGAAACCC', 'AAAGGGCCC'],
    'structure': ['(((...)))', '(((...)))']
})
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')

# Convert Series to JSON
json_str = df['secstruct'].rna.to_json(indent=2)
print("JSON representation:")
print(json_str[:200] + "...")

# Convert back from JSON
structs_restored = df['secstruct'].rna.from_json(json_str)
print(f"\nRestored {len(structs_restored)} structures")
print(f"First structure sequence: {structs_restored.iloc[0].sequence}")
```

## Working with Large Datasets


```python
# Create larger dataset
sequences = ['GGGAAACCC'] * 10
structures = ['(((...)))'] * 10

df = pd.DataFrame({
    'id': range(10),
    'sequence': sequences,
    'structure': structures
})

# Add SecStruct column
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')

# Add statistics
df = df.rna.add_statistics('secstruct')

print(f"DataFrame shape: {df.shape}")
print("\nStatistics summary:")
print(df[['secstruct_num_bp', 'secstruct_gc_content']].describe())
```

## Filtering and Analysis


```python
df = pd.DataFrame({
    'sequence': ['GGGAAACCC', 'AAAGGGCCC', 'CCCGAAAGGG'],
    'structure': ['(((...)))', '(((...)))', '(((...)))']
})
df['secstruct'] = df.rna.from_sequence_structure('sequence', 'structure')
df = df.rna.add_statistics('secstruct')

# Filter by statistics
high_gc = df[df['secstruct_gc_content'] > 0.6]
print(f"Structures with GC content > 60%: {len(high_gc)}")

# Group by statistics
grouped = df.groupby('secstruct_num_bp').size()
print("\nGrouped by number of base pairs:")
print(grouped)
```
