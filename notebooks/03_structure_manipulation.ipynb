# Structure Manipulation

This notebook demonstrates the SequenceStructure-like methods for manipulating RNA structures:
- Slicing
- Insert, replace, remove operations
- Split and join operations
- Immutable container pattern


```python
from rna_secstruct import SecStruct
```

## Slicing


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Slice structure (returns new SecStruct)
sliced = struct[3:6]
print(f"Original: {struct.sequence}, {struct.structure}")
print(f"Sliced [3:6]: {sliced.sequence}, {sliced.structure}")
print(f"\nOriginal unchanged: {struct.sequence}")
print(f"Sliced is new object: {id(sliced) != id(struct)}")
```

## Insert Operations


```python
struct = SecStruct("GGGAAACCC", "(((...)))")
insert_seq = SecStruct("XXX", "...")

# Insert at beginning
result1 = struct.insert(0, insert_seq)
print(f"Insert at 0: {result1.sequence}, {result1.structure}")

# Insert in middle
result2 = struct.insert(3, insert_seq)
print(f"Insert at 3: {result2.sequence}, {result2.structure}")

# Insert at end
result3 = struct.insert(9, insert_seq)
print(f"Insert at 9: {result3.sequence}, {result3.structure}")

# Original unchanged
print(f"\nOriginal unchanged: {struct.sequence}")
```

## Replace Operations


```python
struct = SecStruct("GGGAAACCC", "(((...)))")
replace_seq = SecStruct("XXX", "...")

# Replace in middle
result = struct.replace(replace_seq, 3)
print(f"Replace at 3: {result.sequence}, {result.structure}")
print(f"Original unchanged: {struct.sequence}")
```

## Remove Operations


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Remove middle region
result = struct.remove(3, 6)
print(f"Remove [3:6]: {result.sequence}, {result.structure}")
print(f"Original unchanged: {struct.sequence}")
```

## Subtract Operations


```python
struct = SecStruct("GGGAAACCC", "(((...)))")
sub = SecStruct("AAA", "...")

# Subtract substructure
result = struct.subtract(sub)
print(f"Subtract 'AAA': {result.sequence}, {result.structure}")
print(f"Original unchanged: {struct.sequence}")
```

## Split and Join


```python
# Multi-strand structure
multi_strand = SecStruct("GGG&AAA&CCC", "(((&)))&...")

# Split into individual strands
strands = multi_strand.split_strands()
print(f"Original: {multi_strand.sequence}")
print(f"\nSplit into {len(strands)} strands:")
for i, strand in enumerate(strands):
    print(f"  Strand {i}: {strand.sequence}, {strand.structure}")

# Join strands back together
joined = strands[0].join(strands[1])
print(f"\nJoined first two: {joined.sequence}, {joined.structure}")
```

## Immutable Container Pattern


```python
struct = SecStruct("GGGAAACCC", "(((...)))")
original_id = id(struct)

# All operations return new instances
operations = [
    ("slice", struct[2:7]),
    ("insert", struct.insert(3, SecStruct("X", "."))),
    ("replace", struct.replace(SecStruct("X", "."), 0)),
    ("remove", struct.remove(3, 6)),
    ("subtract", struct.subtract(SecStruct("AAA", "..."))),
]

print("All operations return new instances:")
for op_name, result in operations:
    print(f"  {op_name}: {id(result) != original_id}")

print(f"\nOriginal unchanged: {struct.sequence}")
print(f"Original ID unchanged: {id(struct) == original_id}")
```

## Dictionary and CSV Representations


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Dictionary representation
d = struct.to_dict()
print(f"Dictionary: {d}")

# CSV representation
csv = struct.to_comma_delimited()
print(f"\nCSV: {csv}")
```
