# Search and Analysis

This notebook demonstrates search capabilities and analysis methods:
- Finding substructures
- Pattern matching
- Motif search
- Statistics and analysis


```python
from rna_secstruct import MotifSearchParams, SecStruct
```

## Finding Substructures


```python
struct = SecStruct("GGGAAACCCUUU", "(((...)))((...))")
sub = SecStruct("AAA", "...")

# Find all occurrences
matches = struct.find(sub)
print(f"Structure: {struct.sequence}, {struct.structure}")
print(f"Substructure: {sub.sequence}, {sub.structure}")
print(f"\nFound {len(matches)} matches:")
for start, end in matches:
    print(f"  Position {start}-{end}: {struct.sequence[start:end]}")

# Find with bounds
matches_bounded = struct.find(sub, start=0, end=9)
print(f"\nMatches in range [0:9]: {len(matches_bounded)}")
```

## Sequence Pattern Matching


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Simple sequence search
matches = struct.find_sequence("AAA")
print(f"Found 'AAA' at positions: {matches}")

# Pattern matching with wildcards
matches_wildcard = struct.find_sequence("NNN", allow_wildcards=True)
print(f"\nFound 'NNN' (wildcard) at positions: {matches_wildcard}")

# Without wildcards (literal search)
matches_literal = struct.find_sequence("NNN", allow_wildcards=False)
print(f"Found 'NNN' (literal) at positions: {matches_literal}")
```

## Structure Pattern Matching


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

# Find structure patterns
matches = struct.find_structure("...")
print(f"Found '...' pattern at positions: {matches}")

matches_brackets = struct.find_structure("(((")
print(f"Found '(((' pattern at positions: {matches_brackets}")
```

## Motif Search


```python
struct = SecStruct("GGAAACGAAACGAAACC", "((...)(...)(...))")

# Search by motif type
msp = MotifSearchParams(m_type="HAIRPIN")
hairpins = struct.get_motifs(msp)
print(f"Found {len(hairpins)} hairpins")
for motif in hairpins:
    print(f"  {motif.sequence}, {motif.structure}")

# Search by sequence
msp_seq = MotifSearchParams(sequence="GAAAC")
matches = struct.get_motifs(msp_seq)
print(f"\nFound {len(matches)} motifs with sequence 'GAAAC'")
```

## Enhanced Search Parameters


```python
struct = SecStruct("GGAAACGAAACGAAACC", "((...)(...)(...))")

# Search with multiple criteria
msp = MotifSearchParams(
    m_type="HAIRPIN",
    min_length=4,
    max_length=6,
    has_children=False
)
matches = struct.get_motifs(msp)
print(f"Found {len(matches)} hairpins with length 4-6 and no children")

# Search by strand lengths
junctions = struct.get_motifs_by_strand_lengths([1, 1, 1])
print(f"\nFound {len(junctions)} junctions with strand lengths [1,1,1]")
```

## Statistics and Analysis


```python
struct = SecStruct("GGGAAACCC", "(((...)))")

print(f"Number of base pairs: {struct.get_num_basepairs()}")
print(f"Number of unpaired nucleotides: {struct.get_num_unpaired()}")
print(f"GC content: {struct.get_gc_content():.2%}")
print(f"Helix lengths: {struct.get_helix_lengths()}")
print(f"Number of motifs: {struct.get_num_motifs()}")
```

## Comparison Operations


```python
struct1 = SecStruct("GGGAAACCC", "(((...)))")
struct2 = SecStruct("GGGAAACCC", "(((...)))")
struct3 = SecStruct("AAAGGGCCC", "(((...)))")

# Equality comparison
print(f"struct1 == struct2: {struct1 == struct2}")
print(f"struct1 == struct3: {struct1 == struct3}")

# Structural similarity
similarity = struct1.structural_similarity(struct3)
print(f"\nStructural similarity (struct1 vs struct3): {similarity:.2%}")

# Sequence identity
identity = struct1.sequence_identity(struct3)
print(f"Sequence identity (struct1 vs struct3): {identity:.2%}")
```

## Validation


```python
# Valid structure
valid = SecStruct("GGGAAACCC", "(((...)))")
print(f"Is valid: {valid.is_valid()}")

# Normalize structure (uppercase, T->U)
mixed = SecStruct("gggaaaccc", "(((...)))")
normalized = mixed.normalize()
print(f"\nOriginal: {mixed.sequence}")
print(f"Normalized: {normalized.sequence}")
print(f"T->U conversion: {normalized.sequence.replace('T', 'U') == normalized.sequence}")
```
